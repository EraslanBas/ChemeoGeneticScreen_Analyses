```python
allControls=sc.read_h5ad("/processed_datasets/VCI/ChemoGenetic_H1_Basak/AllControls_scaled.h5ad")
```

```python
import numpy as np
import pandas as pd
import scipy.sparse as sp
import anndata as ad

# Expression matrix
X = allControls.X
if sp.issparse(X):
    X = X.tocsr()

# Grouping labels
samples = allControls.obs["sample"].astype(str).to_numpy()
genes = allControls.var_names.to_numpy()

# Encode samples as categories
cat = pd.Categorical(samples)
codes = cat.codes
sample_names = cat.categories

n_samples = len(sample_names)
n_genes = allControls.n_vars

# Output matrix
pb_mean = np.zeros((n_samples, n_genes), dtype=np.float64)
n_cells = np.bincount(codes)

# Compute mean per sample
for i in range(n_samples):
    mask = codes == i
    if n_cells[i] == 0:
        continue

    if sp.issparse(X):
        pb_mean[i] = np.asarray(X[mask].mean(axis=0)).ravel()
    else:
        pb_mean[i] = X[mask].mean(axis=0)

# Build AnnData
pb_adata = ad.AnnData(
    X=pb_mean,   # dense is fine here; means are rarely sparse
    obs=pd.DataFrame(index=sample_names),
    var=allControls.var.copy()
)

pb_adata.obs["n_cells"] = n_cells

pb_adata

```

```python
from __future__ import annotations
from scipy.stats import fisher_exact
from statsmodels.stats.multitest import multipletests


def enrichment_assignment_in_leiden_per_sample(
    adata,
    *,
    sample_key: str = "sample",
    leiden_key: str = "leiden",
    assign_key: str = "assignment_mixture",
    min_count: int = 5,
    alternative: str = "greater",   # enrichment = OR>1
    fdr_method: str = "fdr_bh",
) -> pd.DataFrame:
    """
    For each sample, test whether each assignment_mixture category is enriched in each leiden category
    using 2x2 Fisher's exact tests (one-vs-rest).

    Returns a tidy DataFrame with odds ratios, p-values, and FDR (per sample).
    """
    obs = adata.obs[[sample_key, leiden_key, assign_key]].copy()

    # basic cleanup
    obs = obs.dropna(subset=[sample_key, leiden_key, assign_key])
    obs[sample_key] = obs[sample_key].astype("string")
    obs[leiden_key] = obs[leiden_key].astype("string")
    obs[assign_key] = obs[assign_key].astype("string")

    results = []

    for sample, df_s in obs.groupby(sample_key, sort=True):
        # contingency table: rows=leiden, cols=assignment
        tab = pd.crosstab(df_s[leiden_key], df_s[assign_key])

        # skip tiny samples
        if tab.values.sum() == 0:
            continue

        pvals = []
        rows = []

        # Precompute totals
        N = int(tab.values.sum())

        for leiden_cat in tab.index:
            for assign_cat in tab.columns:
                a = int(tab.loc[leiden_cat, assign_cat])                # in leiden & in assignment
                if a < min_count:
                    continue

                # one-vs-rest 2x2:
                #          assign_cat   not_assign
                # leiden        a          b
                # not_leiden    c          d
                b = int(tab.loc[leiden_cat, :].sum() - a)
                c = int(tab.loc[:, assign_cat].sum() - a)
                d = N - a - b - c

                # guard (shouldn't happen, but safe)
                if min(a, b, c, d) < 0:
                    continue

                # Fisher exact
                OR, p = fisher_exact([[a, b], [c, d]], alternative=alternative)

                rows.append(
                    dict(
                        sample=sample,
                        leiden=leiden_cat,
                        assignment=assign_cat,
                        a=a, b=b, c=c, d=d,
                        odds_ratio=float(OR) if np.isfinite(OR) else np.inf,
                        p_value=float(p),
                        frac_in_leiden=a / (a + b) if (a + b) > 0 else np.nan,
                        frac_overall=(a + c) / N if N > 0 else np.nan,
                    )
                )
                pvals.append(p)

        if not rows:
            continue

        res_s = pd.DataFrame(rows)

        # FDR correction within sample
        rej, qvals, _, _ = multipletests(res_s["p_value"].values, method=fdr_method)
        res_s["q_value"] = qvals
        res_s["significant_fdr"] = rej

        # optional: simple effect size as log2(OR)
        res_s["log2_odds_ratio"] = np.log2(res_s["odds_ratio"].replace(0, np.nan))

        results.append(res_s)

    if not results:
        return pd.DataFrame(
            columns=[
                "sample", "leiden", "assignment", "a", "b", "c", "d",
                "odds_ratio", "log2_odds_ratio", "p_value", "q_value",
                "significant_fdr", "frac_in_leiden", "frac_overall",
            ]
        )

    out = pd.concat(results, ignore_index=True)

    # nice sorting: strongest + most significant first within each sample
    out = out.sort_values(
        ["sample", "q_value", "p_value", "log2_odds_ratio"],
        ascending=[True, True, True, False],
    )

    return out


# ---------------- Example usage ----------------
# )
# enr.query("q_value < 0.05").head(20)
#
# If you want a matrix per sample (leiden x assignment) of log2 OR:
# mat = enr.pivot_table(index=["sample","leiden"], columns="assignment", values="log2_odds_ratio")
```
