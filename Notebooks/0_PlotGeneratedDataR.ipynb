```r
SampleSheet=read.csv("./../../TextFiles/SampleSheet.csv")
rownames(SampleSheet)=SampleSheet$SampleName
SampleSheet = SampleSheet[SampleSheet$ExperimentRound %in% c("Round2_batch1", "Round2_batch2"),]
```

```r
options(repr.plot.width = 13, repr.plot.height = 11)

sampleCosineDistances=read.csv("./TextFiles/cosine_distances_10000cells.csv", 
                       row.names = 1)
colnames(sampleCosineDistances)=rownames(sampleCosineDistances)
sampleCosineDistances = sampleCosineDistances[rownames(sampleCosineDistances) %in% SampleSheet$SampleName,]
sampleCosineDistances = sampleCosineDistances[,colnames(sampleCosineDistances) %in% SampleSheet$SampleName,]

SampleSheet <- SampleSheet[rownames(sampleCosineDistances),]
annots=copy(SampleSheet)
annots$SampleName=NULL
tmp=copy(sampleCosineDistances)
#tmp[tmp>0.6]=0.6
tmp
pheatmap(
  tmp,
  # cluster_cols=FALSE,
  # cluster_rows=FALSE,
  number_color = "black",
  fontsize_number = 10,
  fontsize_col=15,fontsize_row=15,
  annotation_row=annots,
  annotation_col=annots,
  clustering_method = "ward.D2",
  color = colorRampPalette(c("blue", "white","yellow","orange", "red"))(100),
  main = "Cosine distances between drug NTC cells"
)
```

```r
options(repr.plot.width = 11, repr.plot.height = 11)

Jaccard_perts=read.csv("FDR_matrices/jaccard_outputs/DMSO_round2_perturbations.csv", 
                       row.names = 1)
baselineNSigGenes = Jaccard_perts["baseline_sig_n_genes"]
Jaccard_perts$baseline_sig_n_genes=NULL

km2 <- pheatmap(
  Jaccard_perts,
  number_color = "black",
  fontsize_number = 10,
  fontsize_col=15,fontsize_row=1,
  clustering_method = "ward.D2",
  color = colorRampPalette(c("blue", "white","yellow","orange", "red"))(100),
  main = "Jaccard similarity between perturbations"
)
rowOrder=rownames(Jaccard_perts)[km2$tree_row$order]
```

```r
options(repr.plot.width = 1.5, repr.plot.height = 11)

baselineNSigGenes[baselineNSigGenes>2000]=2000
pheatmap(
  baselineNSigGenes[rowOrder,],
  number_color = "black",
  fontsize_number = 10,
  fontsize_col=15,fontsize_row=15,
  clustering_method = "ward.D2",
  color = colorRampPalette(c("blue", "white", "red"))(100),
  cluster_rows=FALSE,
  cluster_cols=FALSE
  
)
```

```r
options(repr.plot.width = 11, repr.plot.height = 11)

Jaccard_genes=read.csv("FDR_matrices/jaccard_outputs/DMSO_round2_genes.csv", 
                       row.names = 1)
baseline_sig_n_perts = Jaccard_genes["baseline_sig_n_perts"]
Jaccard_genes$baseline_sig_n_perts=NULL

km2 <- pheatmap(
  Jaccard_genes,
  number_color = "black",
  fontsize_number = 10,
  fontsize_col=15,fontsize_row=1,
  clustering_method = "ward.D",
  color = colorRampPalette(c("blue", "white","yellow","orange", "red"))(100),
  main = "Jaccard similarity between genes"
)
rowOrder=rownames(Jaccard_genes)[km2$tree_row$order]
```

```r
options(repr.plot.width = 1.5, repr.plot.height = 11)

baseline_sig_n_perts[baseline_sig_n_perts>400]=400
pheatmap(
  baseline_sig_n_perts[rowOrder,],
  number_color = "black",
  fontsize_number = 10,
  fontsize_col=15,fontsize_row=15,
  clustering_method = "ward.D2",
  color = colorRampPalette(c("blue", "white", "red"))(100),
  cluster_rows=FALSE,
  cluster_cols=FALSE
  
)
```
